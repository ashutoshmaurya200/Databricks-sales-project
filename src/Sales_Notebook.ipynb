```python
from pyspark.sql.functions import *

data=[
    (1,"Laptop",50000),
    (2,"Mobile",30000),
    (3,"Tablet",20000),
    (4,"Laptop",60000),
    (5,"Mobile",40000)
]

columns =["id","product","price"]
df =spark.createDataFrame(data,columns)
result =df.withColumn("Tax",(col("price")*0.28))
display(result)
```
